# 🧹 Data Cleaning and Preprocessing

**Dataset:** Stock Prices  
**Tools:** Python, pandas

---

We work with a raw dataset that contains:
- Missing values
- Duplicate records
- Inconsistent data formats

The goal is to clean and prepare the dataset for analysis.

---

## Step 1: Load the Dataset

```python
import pandas as pd

df = pd.read_csv("2) Stock Prices Data Set.csv")
df.head()
```

---
## Step 2: Explore the Data

Check the structure, missing values, and duplicates before doing anything.

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

---
## Step 3: Handle Missing Values

Fill missing values in numeric columns using the column mean.

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

---
## Step 4: Remove Duplicate Rows

```python
df.drop_duplicates(inplace=True)
```

---
## Step 5: Standardize Data Formats

### 5.1 — Parse the date column

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['date'].isnull().sum()
```

### 5.2 — Normalize text columns

Convert all string columns to uppercase and remove extra whitespace.

```python
for col in df.select_dtypes(include='object'):
    df[col] = df[col].str.upper().str.strip()
```

---
## Step 6: Validate the Cleaned Data

Re-run the same checks to confirm everything is resolved.

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```
