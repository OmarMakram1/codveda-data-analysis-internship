## 2️⃣ Load the Dataset

## 4️⃣ Data Visualization

In this section, we create several visualizations to explore the dataset from different perspectives.

These plots help us:
- understand correlations between variables,
- study the distribution of important features,
- and identify relationships between housing characteristics and house prices.

### 4.1 🔥 Correlation Matrix

The correlation matrix shows the strength and direction of the relationship between numerical variables in the dataset.

This visualization helps identify:
- strongly related features,
- variables that may influence house prices,
- and potential multicollinearity between predictors.

```python
plt.figure(figsize=(14, 10))
sns.heatmap(df.corr(),cmap='bwr',
            center=0, annot=True)
plt.title('Correlation Matrix')
plt.savefig("Correlation Matrix.png", dpi=300, bbox_inches='tight')

plt.show()
```

### 4.2 📈 Scatter Plot: Rooms vs Median Home Value

This scatter plot examines the relationship between the average number of rooms (**RM**) and the median home value (**MEDV**).

A regression line is added to highlight the overall trend and show whether home prices tend to increase as the number of rooms increases.

```python
r = df["RM"].corr(df["MEDV"])

fig, ax = plt.subplots(figsize=(10, 7))

sns.regplot(
    data=df, x="RM", y="MEDV",
    scatter_kws={"alpha": 0.4, "s": 40, "color": "#378ADD"},
    line_kws={"color": "#E24B4A", "linewidth": 2},
    ax=ax
)

ax.set_title("Rooms vs. Median Home Value ", fontsize=15, pad=15, fontweight="bold")
ax.set_xlabel("RM – Average number of rooms", fontsize=12)
ax.set_ylabel("MEDV – Median value ", fontsize=12)

sns.despine()
plt.tight_layout()
plt.savefig("Rooms vs. Median Home Value.png", dpi=300, bbox_inches='tight')
plt.show()
```

### 4.3 📊 Distribution of Median Home Value

This histogram shows how the target variable (**MEDV**) is distributed across the dataset.

It helps us understand:
- the most common home values,
- the spread of the data,
- and whether the distribution is symmetric or skewed.

```python
plt.figure(figsize=(10, 7))


sns.histplot(
    data=df, x="MEDV", kde=True,
    color="#800e13",
    edgecolor="white",
    linewidth=0.6,
    alpha=0.75,
)
plt.title("Distribution of Median Home Value", fontsize=15, pad=15, fontweight="bold")
sns.despine()
plt.savefig("Distribution of Median Home Value.png", dpi=300, bbox_inches='tight')

plt.show()
```

### 4.4 📉 Scatter Plot: LSTAT vs Median Home Value

This scatter plot explores the relationship between **LSTAT** and **MEDV**.

- **LSTAT** represents the percentage of lower socio-economic status in the neighborhood.
- **MEDV** represents the median home value.

This visualization helps show how house prices change as the LSTAT percentage increases.

```python


plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=df, x="LSTAT", y="MEDV",
    s=40, alpha=0.55, edgecolor="white", linewidth=0.5
)

sns.regplot(
    data=df, x="LSTAT", y="MEDV",
    scatter=False, lowess=True,
    line_kws={"color": "#D81B60", "lw": 2}
)
plt.title("LSTAT vs MEDV", loc="center", weight="bold")
plt.xlabel("LSTAT — Percentage of lower socio-economic status")
plt.ylabel("MEDV — Median home value ")

sns.despine()
plt.tight_layout()
plt.savefig("LSTAT vs MEDV.png", dpi=300, bbox_inches='tight')

plt.show()
```

### 4.6 📊 Bar Plot: Average House Price by Charles River Proximity

This bar chart compares the average median home value (**MEDV**) for houses that are near the Charles River and those that are not.

It provides a simple categorical comparison between the two groups.

```python
plt.figure(figsize=(8,5))

avg_price_by_chas = df.groupby('CHAS')['MEDV'].mean()

avg_price_by_chas.plot(kind='bar', color=['#800e13', '#ffba08'])

plt.title('Average House Price by Charles River Proximity')
plt.xlabel('CHAS (0 = No, 1 = Yes)')
plt.ylabel('Average MEDV')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('Average_House_Price_by_CHAS.png', dpi=300, bbox_inches='tight')
plt.show()
```

### 4.7 📈 Line Chart: Median Home Value Across Records

This line chart displays the median home value (**MEDV**) across a subset of records in the dataset.

It helps visualize how the target variable changes from one observation to another and provides a simple trend view.

```python
plt.figure(figsize=(12,6))

plt.plot(df.index[:50], df['MEDV'][:50], marker='o', color='#800e13', label='MEDV')

plt.title('Line Chart of Median Home Value (First 50 Records)')
plt.xlabel('Index')
plt.ylabel('MEDV')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('Line_Chart_MEDV_First_50.png', dpi=300, bbox_inches='tight')
plt.show()
```
